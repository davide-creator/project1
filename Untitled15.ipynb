```python
import os
import pandas as pd
import win32com.client as win32

def parse_config(config_file_path):
    """
    Legge il file di testo di configurazione (config.txt)
    e ne restituisce un dizionario {chiave:valore}.
    Ignora righe vuote o che iniziano con '#'.
    """
    config = {}
    with open(config_file_path, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if not line or line.startswith('#'):
                continue
            if '=' in line:
                key, value = line.split('=', 1)
                config[key.strip()] = value.strip()
    return config

def get_outlook_attachment(outlook_folder_name):
    """
    Apre Outlook e cerca la prima mail nella cartella `outlook_folder_name`.
    Salva in locale il primo allegato Excel trovato.
    Ritorna il percorso del file salvato o None se non trova nulla.
    """
    outlook = win32.Dispatch("Outlook.Application").GetNamespace("MAPI")

    # Se devi accedere a un account specifico
    #   account = outlook.Folders["indirizzo_email@qualcosa.it"]
    #   folder = account.Folders[outlook_folder_name]
    # Altrimenti, spesso il primo .Folders.Item(1) è la root del primo account
    folder = outlook.Folders.Item(1).Folders[outlook_folder_name]

    items = folder.Items
    if len(items) == 0:
        print("Nessuna email trovata nella cartella:", outlook_folder_name)
        return None

    mail = items[0]  # ipotizziamo di prendere la prima mail
    attachments = mail.Attachments
    if attachments.Count == 0:
        print("Nessun allegato trovato nella mail.")
        return None

    attachment = attachments.Item(1)  # primo allegato
    save_path = os.path.join(os.getcwd(), attachment.FileName)
    attachment.SaveAsFile(save_path)
    return save_path

def read_attached_excel(file_path):
    """
    Legge l'Excel allegato (senza header) perché la prima colonna (A) contiene i nomi delle variabili.
    Le colonne successive (B, C, D, ecc.) contengono i valori per ogni lancio.
    Ritorna un DataFrame con .shape = (num_righe, num_colonne).
    Non usiamo intestazioni, quindi header=None.
    """
    df = pd.read_excel(file_path, header=None)
    return df

def read_pricing_grid(pricing_grid_path):
    """
    Legge la Pricing Grid, che presumiamo abbia un header "umano" (riga 0 con i nomi delle colonne).
    Ritorna un DataFrame con le colonne nominative, ad es.:
       "Fund Company", "Sub-Fund Name", "Share Class", "ISIN", "FSF", "TAX", "ADV Rate", ...
    """
    df = pd.read_excel(pricing_grid_path)
    return df

def main():
    # 1) Leggo la config
    config_file = "config.txt"  # Cambia se il tuo file di config ha un altro nome
    config = parse_config(config_file)

    # Estraggo i valori di interesse dal dizionario 'config'
    outlook_folder_name = config.get("OutlookFolderName")
    pricing_grid_path = config.get("PricingGridPath")
    output_draft_path = config.get("OutputDraftPath")

    col_pg_fund_company = config.get("COL_PG_FUND_COMPANY", "Fund Company")
    col_pg_sub_fund = config.get("COL_PG_SUB_FUND", "Sub-Fund Name")
    col_pg_share_class = config.get("COL_PG_SHARE_CLASS", "Share Class")
    col_pg_isin = config.get("COL_PG_ISIN", "ISIN")
    col_pg_fsf = config.get("COL_PG_FSF", "FSF")
    col_pg_tax = config.get("COL_PG_TAX", "TAX")
    col_pg_adv = config.get("COL_PG_ADV_RATE", "ADV Rate")

    # Controlli di base
    if not outlook_folder_name or not pricing_grid_path or not output_draft_path:
        print("ERRORE: mancano parametri fondamentali nel file di config.")
        return

    # 2) Cerco il file Excel allegato in Outlook
    attached_file = get_outlook_attachment(outlook_folder_name)
    if not attached_file:
        print("ERRORE: impossibile procedere senza allegato.")
        return

    print("Allegato salvato in:", attached_file)

    # 3) Leggo il file Excel allegato (orientato su colonne)
    df_attached = read_attached_excel(attached_file)
    # df_attached.iloc[:, 0] => colonna con i NOMI delle variabili (es: "Fund Company", "Sub-Fund Name", "ISIN", "FSF", ecc.)
    # df_attached.iloc[:, 1..N] => ognuna è un lancio.

    num_rows, num_cols = df_attached.shape
    if num_cols < 2:
        print("ATTENZIONE: c'è solo la colonna con i nomi delle variabili. Nessun lancio rilevato.")
        return

    # Ricavo la lista di variabili (prima colonna)
    variable_names = df_attached.iloc[:, 0].tolist()  # righe della colonna 0

    # 4) Creiamo una lista con i "dizionari" di variabili-valori per ciascun lancio
    launches = []
    for col_idx in range(1, num_cols):  # dalla seconda colonna fino all'ultima
        launch_data = {}
        for row_idx in range(num_rows):
            var_name = variable_names[row_idx]
            var_value = df_attached.iloc[row_idx, col_idx]
            launch_data[var_name] = var_value
        launches.append(launch_data)

    # 5) Leggo la Pricing Grid
    df_pricing = read_pricing_grid(pricing_grid_path)

    # 6) Per ogni lancio, confronto con la Pricing Grid
    results = []
    errors = []

    for i, launch in enumerate(launches, start=1):
        # Estraggo i campi che servono (adatta i nomi di variabili in base ai tuoi reali):
        fund_company = launch.get("Fund Company", "")
        sub_fund = launch.get("Sub-Fund Name", "")
        share_class = launch.get("Share Class", "")
        isin = launch.get("ISIN", "")
        fsf = launch.get("FSF", "")
        tax = launch.get("TAX", "")
        adv = launch.get("ADV Rate", "")

        # Filtro la Pricing Grid (adatta se i tuoi campi chiave sono diversi)
        subset = df_pricing[
            (df_pricing[col_pg_fund_company] == fund_company) &
            (df_pricing[col_pg_sub_fund] == sub_fund) &
            (df_pricing[col_pg_share_class] == share_class)
        ]

        if subset.empty:
            errors.append(
                f"[Lancio {i}] Nessun record trovato in Pricing Grid per: "
                f"Fund Company={fund_company}, Sub-Fund={sub_fund}, ShareClass={share_class}"
            )
        else:
            # Ad esempio, controlliamo se FSF e TAX (nella Pricing Grid) corrispondono ai valori riportati nell’allegato
            # (oppure se hai logiche diverse, adattale).
            pg_fsf_vals = subset[col_pg_fsf].unique()
            pg_tax_vals = subset[col_pg_tax].unique()
            pg_adv_vals = subset[col_pg_adv].unique()

            # Se nella Pricing Grid ci sono più valori diversi, potresti considerarlo un errore
            if len(pg_fsf_vals) > 1:
                errors.append(f"[Lancio {i}] Incoerenza FSF in Pricing Grid: {pg_fsf_vals}")
            if len(pg_tax_vals) > 1:
                errors.append(f"[Lancio {i}] Incoerenza TAX in Pricing Grid: {pg_tax_vals}")
            if len(pg_adv_vals) > 1:
                errors.append(f"[Lancio {i}] Incoerenza ADV Rate in Pricing Grid: {pg_adv_vals}")

            # Se vuoi anche controllare la corrispondenza con i valori provenienti dall'allegato, fallo:
            if len(pg_fsf_vals) == 1:
                pg_fsf = pg_fsf_vals[0]
                if fsf != pg_fsf:
                    errors.append(f"[Lancio {i}] FSF allegato={fsf} ma PricingGrid={pg_fsf}")

            if len(pg_tax_vals) == 1:
                pg_tax = pg_tax_vals[0]
                if tax != pg_tax:
                    errors.append(f"[Lancio {i}] TAX allegato={tax} ma PricingGrid={pg_tax}")

            if len(pg_adv_vals) == 1:
                pg_adv = pg_adv_vals[0]
                if adv != pg_adv:
                    errors.append(f"[Lancio {i}] ADV allegato={adv} ma PricingGrid={pg_adv}")

        # Salvo i dati essenziali per la bozza di email
        results.append({
            "Lancio": i,
            "Fund Company": fund_company,
            "Sub-Fund Name": sub_fund,
            "Share Class": share_class,
            "ISIN": isin,
            "FSF": fsf,
            "TAX": tax,
            "ADV Rate": adv
        })

    # 7) Genero il file di testo (bozza di email)
    with open(output_draft_path, "w", encoding="utf-8") as out:
        out.write("Ciao,\n\nEcco i dati estratti dall'allegato e i relativi controlli:\n\n")
        out.write("Lancio\tFund Company\tSub-Fund Name\tShare Class\tISIN\tFSF\tTAX\tADV\n")
        for r in results:
            out.write(
                f"{r['Lancio']}\t{r['Fund Company']}\t{r['Sub-Fund Name']}\t"
                f"{r['Share Class']}\t{r['ISIN']}\t{r['FSF']}\t{r['TAX']}\t{r['ADV Rate']}\n"
            )

        if errors:
            out.write("\nErrori riscontrati:\n")
            for e in errors:
                out.write(f" - {e}\n")
        else:
            out.write("\nNessun errore riscontrato.\n")

        out.write("\nCordiali saluti,\nNome Cognome\n")

    print("Bozza di email generata in:", output_draft_path)

if __name__ == "__main__":
    main()
```
