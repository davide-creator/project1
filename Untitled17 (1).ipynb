```python
import os
import re
from datetime import datetime
from openpyxl import Workbook, load_workbook
from openpyxl.utils import (
    column_index_from_string,
    get_column_letter
)


def leggi_variabili_da_file(percorso_file: str):
    """Legge righe non vuote e non commentate (#...) da un file di testo."""
    variabili = []
    if os.path.exists(percorso_file):
        with open(percorso_file, "r", encoding="utf-8") as f:
            for line in f:
                line = line.strip()
                if not line or line.startswith('#'):
                    continue
                variabili.append(line)
    return variabili


def copia_sheet_con_formule(ws_origine, ws_destinazione):
    """
    Copia TUTTE le celle (valori e formule) dallo sheet di origine
    allo sheet di destinazione, mantenendo la stessa posizione
    (riga/colonna). Non copia stili, formati, dimensioni colonne, ecc.

    ws_origine: Worksheet di openpyxl da cui copiare
    ws_destinazione: Worksheet di openpyxl dove incollare
    """
    max_row = ws_origine.max_row
    max_col = ws_origine.max_column

    for r in range(1, max_row + 1):
        for c in range(1, max_col + 1):
            cell_origine = ws_origine.cell(row=r, column=c)
            cell_dest = ws_destinazione.cell(row=r, column=c)

            # Copia semplicemente il 'value' così com'è.
            # Se era una formula tipo "=SOMMA(A1:B1)", rimane uguale.
            cell_dest.value = cell_origine.value


def shift_cell_references(formula, row_shift=1, col_shift=0):
    """
    Esegue lo shift dei riferimenti di cella nelle formule.
    - row_shift: +1 sposta in basso di 1 riga, ecc.
    - col_shift: se vuoi spostare anche le colonne, regolarlo a +1 o altro.

    La regex cattura riferimenti come:
        'Foglio'!A1, A1, $A1, A$1, $A$1
    e anche range tipo A1:B10.
    Se la riga è assoluta (es. $4), NON viene spostata.
    """
    cell_ref_pattern = re.compile(
        r"((?:'[^']+'!)?)"           # Gruppo (opzionale) per il nome dello sheet + '!'
        r"(\$?[A-Z]{1,3})"           # Colonna, es: A, B, $AA, ...
        r"(\$?\d+)"                  # Riga, es: 10, $10, ...
    )

    def replace_match(match):
        sheet_part = match.group(1)   # Es. "'Sheet1'!" o vuoto
        col_part   = match.group(2)   # Es. "A" o "$A"
        row_part   = match.group(3)   # Es. "10" o "$10"

        # ---- Gestione colonna ----
        col_abs = col_part.startswith('$')
        col_letters = col_part[1:] if col_abs else col_part  # togli $ se presente
        # Convertiamo in indice (A->1, B->2, ecc.):
        col_idx = column_index_from_string(col_letters)
        if not col_abs:  # solo se non è assoluta
            col_idx += col_shift
        # Convertiamo di nuovo in lettere:
        new_col_letters = get_column_letter(col_idx)
        if col_abs:
            new_col_letters = f'${new_col_letters}'

        # ---- Gestione riga ----
        row_abs = row_part.startswith('$')
        row_number = int(row_part[1:] if row_abs else row_part)
        if not row_abs:
            row_number += row_shift
        new_row_part = f'${row_number}' if row_abs else str(row_number)

        return f"{sheet_part}{new_col_letters}{new_row_part}"

    # Sostituzione su formula (eventuali range A1:B10 inclusi)
    if formula.startswith('='):
        formula_no_equals = formula[1:]
    else:
        formula_no_equals = formula

    new_formula = cell_ref_pattern.sub(lambda m: replace_match(m), formula_no_equals)
    return "=" + new_formula


def applica_shift_alle_nuove_righe(ws, first_new_row, last_new_row):
    """
    Per ogni riga nel range [first_new_row .. last_new_row],
    copia la formula dalla riga precedente, ma effettua lo shift
    dei riferimenti di +1 riga.
    """
    for row_idx in range(first_new_row, last_new_row + 1):
        row_above = row_idx - 1
        if row_above < 1:
            continue

        for col_idx in range(1, ws.max_column + 1):
            cell_above = ws.cell(row=row_above, column=col_idx)
            cell_new   = ws.cell(row=row_idx,   column=col_idx)

            if cell_above.data_type == 'f':
                # Esempio formula: =SUM(A4:B4)
                old_formula = cell_above.value  # string
                new_formula = shift_cell_references(old_formula, row_shift=1, col_shift=0)
                cell_new.value = new_formula


def main():
    # 1) Leggere variabili (se servono)
    variabili = leggi_variabili_da_file("variabili.txt")
    print("Variabili:", variabili)

    # 2) Aprire il file sorgente SENZA mai salvarlo (non lo tocchiamo)
    #    data_only=False per preservare le formule
    source_path = "file_originale.xlsx"  # <-- metti il nome del file vero
    source_wb = load_workbook(source_path, data_only=False)

    ws_closed_src = source_wb["closed"]
    ws_main_src   = source_wb["main"]

    # 3) Creare un NUOVO workbook per non modificare l'originale
    dest_wb = Workbook()

    # Di default c'è un sheet chiamato "Sheet", lo rimuoviamo se vogliamo
    default_sheet = dest_wb.active
    dest_wb.remove(default_sheet)

    # Creiamo un nuovo sheet "main" nel workbook di destinazione
    ws_main_dest = dest_wb.create_sheet("main")

    # 4) Copiare TUTTO lo sheet "main" (esistente) -> "main" destinazione
    copia_sheet_con_formule(ws_main_src, ws_main_dest)

    # 5) Identificare le righe in "closed" degli ultimi due anni
    anno_corrente = datetime.now().year
    anni_validi = [anno_corrente, anno_corrente - 1]

    # Colonna BM
    colonna_BM_idx = column_index_from_string("BM")  # BM -> 65

    righe_da_copiare = []
    # Partiamo da riga 2 se la riga 1 è intestazione
    for row in ws_closed_src.iter_rows(min_row=2, values_only=False):
        cell_BM = row[colonna_BM_idx - 1]
        if not cell_BM.value:
            continue

        # Proviamo a capire se è datetime
        if isinstance(cell_BM.value, datetime):
            anno_data = cell_BM.value.year
        else:
            # Se è stringa es: "15-03-2025", tentiamo un parse manuale
            # Se il formato è "GG-MM-YYYY"
            try:
                dt = datetime.strptime(str(cell_BM.value), "%d-%m-%Y")
                anno_data = dt.year
            except:
                continue  # se non riesce a fare parse, saltiamo

        if anno_data in anni_validi:
            righe_da_copiare.append(row)

    # Se non abbiamo righe da copiare, finiamo subito
    if not righe_da_copiare:
        print("Nessuna riga con date degli ultimi due anni trovata in 'closed'.")
    else:
        print(f"Trovate {len(righe_da_copiare)} righe da copiare in 'main'.")

        # 6) Incollare le righe in coda a 'main' (nel workbook di destinazione)
        first_empty_row = ws_main_dest.max_row + 1
        for i, row_cells in enumerate(righe_da_copiare, start=0):
            target_row = first_empty_row + i
            for col_idx, cell_src in enumerate(row_cells, start=1):
                ws_main_dest.cell(row=target_row, column=col_idx).value = cell_src.value

        # 7) Eseguire lo shift delle formule sulle nuove righe incollate
        last_new_row = first_empty_row + len(righe_da_copiare) - 1
        applica_shift_alle_nuove_righe(ws_main_dest, first_empty_row, last_new_row)

    # 8) Salvare il nuovo file. L'originale non viene toccato.
    dest_path = "Pricing Grid.xlsx"  # <-- Nome del file di output
    dest_wb.save(dest_path)
    print(f"Fatto! Salvato in '{dest_path}'. L'originale '{source_path}' è rimasto invariato.")


if __name__ == "__main__":
    main()
```
