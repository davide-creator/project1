```python
import os
from datetime import datetime
import re
from openpyxl import load_workbook
from openpyxl.utils import column_index_from_string


def leggi_variabili_da_file(percorso_file: str):
    """
    Legge un file di testo che contiene nomi di variabili (uno per riga),
    ignorando righe vuote e righe di commento (che iniziano con '#').
    Restituisce una lista di stringhe (le variabili).
    """
    variabili = []
    if os.path.exists(percorso_file):
        with open(percorso_file, 'r', encoding='utf-8') as f:
            for line in f:
                line = line.strip()
                if not line or line.startswith('#'):
                    continue
                variabili.append(line)
    return variabili


def shift_cell_references(formula, row_shift=1, col_shift=0):
    """
    Esegue lo shift dei riferimenti di cella (sia singole celle es. A4, sia range A4:B10)
    in una formula Excel. Utilizza un'espressione regolare per catturare i riferimenti
    e aggiornarli di conseguenza.

    row_shift: di quanto spostare la riga (es. +1 se sto copiando la formula in una riga più in basso).
    col_shift: di quanto spostare la colonna (qui impostato a 0 se non serve).

    Gestisce:
      - Forme come: A1, $A1, A$1, $A$1
      - Sheet facoltativo: 'SheetName'!A1
      - Riferimenti di range: A1:B2, 'SheetName'!A1:'SheetName'!B2 (nella formula).

    Non gestisce riferimenti di tabelle strutturate o nomi di range complessi.
    Non gestisce riferimenti assoluti se il simbolo `$` è presente (verrà lasciato invariato).
    Esempio: $A$4 resta $A$4, non subisce lo shift della riga.
    """

    # Regex per trovare riferimenti di cella o range, con eventuale nome di sheet
    # Gruppi di interesse:
    #   1) sheet (facoltativo), es: 'Foglio 1'!
    #   2) colonna (es. A, B, $AA)
    #   3) riga (es. 10, $10)
    # Può apparire anche un range separato da ":", per esempio "A1:B10".
    # Faremo il replace su entrambe le parti se c'è il ":".
    cell_ref_pattern = re.compile(
        r"((?:'[^']+'!)?)"           # Gruppo "sheet" (facoltativo, 'SheetName'!)
        r"(\$?[A-Z]{1,3})"           # Gruppo "colonna" (es. A, B, $AA) - 1-3 lettere
        r"(\$?\d+)"                  # Gruppo "riga" (es. 12, $12)
    )

    def replace_match(match):
        sheet_part = match.group(1)  # Es: "'Foglio1'!"
        col_part   = match.group(2)  # Es: "A" o "$A"
        row_part   = match.group(3)  # Es: "10" o "$10"

        # Gestione colonna
        # Se col_part ha un '$', non shiftiamo (colonna assoluta).
        # Altrimenti potremmo calcolare colonna -> indice -> sommare col_shift -> ritrasformare
        # Qui facciamo col_shift=0, ma lasciamo in esempio come si potrebbe fare:

        col_absolute = col_part.startswith('$')
        if col_absolute:
            # Rimuovo il simbolo '$' per calcolare l'indice, ma NON lo toglierò dal risultato finale
            col_letters = col_part[1:]  # es. "A"
        else:
            col_letters = col_part

        # Se ti interessa lo shift di colonna, calcola:
        # col_index = column_index_from_string(col_letters)  # A -> 1, B -> 2 ...
        # new_col_index = col_index + col_shift
        # # Converti indietro, es. 1 -> A
        # new_col_letters = get_column_letter(new_col_index)
        # Se col_absolute = True, rimetto il $
        # if col_absolute:
        #     col_result = f'${new_col_letters}'
        # else:
        #     col_result = new_col_letters
        #
        # In questo esempio lasciamo col_shift=0, quindi col_result = col_part
        col_result = col_part

        # Gestione riga
        # Se row_part ha il '$', non shiftiamo (riga assoluta),
        # altrimenti effettuiamo row += row_shift
        row_absolute = row_part.startswith('$')
        if row_absolute:
            # Rimuovo il '$' per convertire in int
            row_number = int(row_part[1:])
        else:
            row_number = int(row_part)

        # Applica shift di riga solo se non è assoluta
        if not row_absolute:
            row_number += row_shift

        # Ricostruisco il riferimento di riga
        if row_absolute:
            row_result = f'${row_number}'
        else:
            row_result = str(row_number)

        # Riassemblo: 'Foglio1'! + colonna + riga
        return f"{sheet_part}{col_result}{row_result}"

    # Ora dobbiamo gestire il caso di range (es. A1:B2).
    # Possiamo semplicemente fare re.sub su tutta la formula,
    # e si occuperà di sostituire sia la parte prima che quella dopo i due punti.
    new_formula = cell_ref_pattern.sub(lambda m: replace_match(m), formula)

    return new_formula


def applica_formule_nuove_righe_con_shift(ws, numero_righe_aggiunte):
    """
    Copia le formule dalla riga precedente, ma fa lo shift dei riferimenti di riga
    all'interno delle formule: es. se la formula era =SUM(A4:B4),
    nella riga sotto diventa =SUM(A5:B5).

    Funziona solo se le formule fanno riferimento alle righe in modo "relativo"
    (cioè senza $ sulla riga). Se una riga è assoluta ($4), non verrà shiftata.
    """
    if numero_righe_aggiunte <= 0:
        return  # Nessuna riga nuova, niente da fare

    # L'ultima riga del worksheet dopo la copia è ws.max_row
    # Quindi le "nuove" righe occupano da (ws.max_row - numero_righe_aggiunte + 1) a ws.max_row
    first_new_row = ws.max_row - numero_righe_aggiunte + 1

    for row_idx in range(first_new_row, first_new_row + numero_righe_aggiunte):
        # Riga immediatamente sopra
        row_above = row_idx - 1
        if row_above < 1:
            continue

        for col_idx in range(1, ws.max_column + 1):
            cell_above   = ws.cell(row=row_above, column=col_idx)
            cell_current = ws.cell(row=row_idx,   column=col_idx)

            if cell_above.data_type == 'f':
                old_formula = cell_above.value  # Es. "=SUM(A4:B4)"
                # Rimuoviamo l'uguale iniziale se presente
                if old_formula.startswith('='):
                    old_formula = old_formula[1:]

                # Eseguiamo lo shift di +1 riga
                new_formula = shift_cell_references(old_formula, row_shift=1, col_shift=0)

                # Rimettiamo l'uguale
                cell_current.value = f"={new_formula}"


def copia_righe(ws_origine, ws_destinazione, righe_da_copiare):
    """
    Copia interamente le righe (tutte le colonne) da ws_origine a ws_destinazione.
    `righe_da_copiare` è una lista di `Row` (tuple di celle) provenienti da ws_origine.
    Restituisce il numero di righe effettivamente copiate.
    """
    # Trova la prossima riga vuota in ws_destinazione
    first_empty_row = ws_destinazione.max_row + 1

    for row_offset, row in enumerate(righe_da_copiare):
        new_row_idx = first_empty_row + row_offset
        for col_idx, cell in enumerate(row, start=1):
            # Copia semplicemente il "value" della cella (senza stile/grafica)
            ws_destinazione.cell(row=new_row_idx, column=col_idx).value = cell.value

    return len(righe_da_copiare)


def main():
    # 1) Leggere le variabili da un file di testo (se serve).
    variabili = leggi_variabili_da_file("variabili.txt")
    print("Variabili lette:", variabili)

    # 2) Aprire il file excel sorgente
    wb = load_workbook("file.xlsx")  # <-- Cambia in base al tuo file
    ws_closed = wb["closed"]
    ws_main = wb["main"]

    # 3) Individuare la colonna BM
    # Se sai già che "BM" = colonna 65, puoi fare:
    colonna_BM_idx = column_index_from_string("BM")

    # 4) Determinare anno corrente e filtrare le righe
    anno_corrente = datetime.now().year
    anni_valori_validi = [anno_corrente, anno_corrente - 1]

    righe_da_copiare = []
    for row in ws_closed.iter_rows(min_row=2, values_only=False):
        cell_BM = row[colonna_BM_idx - 1]  # in Python l'indice parte da 0
        if cell_BM.value is None:
            continue

        if isinstance(cell_BM.value, datetime):
            anno_data = cell_BM.value.year
            if anno_data in anni_valori_validi:
                righe_da_copiare.append(row)
        else:
            # Se non è datetime, potresti tentare un parse
            pass

    print(f"Righe da copiare: {len(righe_da_copiare)}")

    # 5) Copia le righe nello sheet main
    num_righe_copiate = copia_righe(ws_closed, ws_main, righe_da_copiare)

    # 6) Copia e shift delle formule per le nuove righe
    applica_formule_nuove_righe_con_shift(ws_main, num_righe_copiate)

    # 7) Salvataggio in un nuovo file
    wb.save("Pricing Grid.xlsx")
    print("Fatto. Salvato in 'Pricing Grid.xlsx'.")


if __name__ == "__main__":
    main()
```
