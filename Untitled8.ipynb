```python
import os
import re
import shutil
import zipfile
import glob

import win32com.client
import pandas as pd
from PyPDF2 import PdfReader, PdfWriter

# Se disponibile, per file Office (doc, xls, …)
try:
    import msoffcrypto
except ImportError:
    msoffcrypto = None


# ============================================
# FUNZIONI DI SUPPORTO
# ============================================

def read_config(excel_path):
    """
    Legge il file Excel di configurazione e restituisce un dizionario
    con le coppie parametro/valore.
    
    Viene utilizzata la prima colonna per le chiavi e la seconda per i valori.
    """
    df = pd.read_excel(excel_path)
    print("Colonne lette:", df.columns.tolist())  # Debug: controlla quali intestazioni sono lette
    config = {}
    # Usa la prima colonna (df.columns[0]) per i nomi delle variabili
    # e la seconda (df.columns[1]) per i valori
    col_key = df.columns[0]
    col_val = df.columns[1]
    for _, row in df.iterrows():
        key = str(row[col_key]).strip()
        value = row[col_val]
        if isinstance(value, str):
            value = value.strip()
        config[key] = value
    print("Configurazione letta:", config)
    return config


def get_processed_email_ids(processed_ids_path):
    """
    Legge il file di testo contenente gli ID delle email già processate.
    Restituisce un set di stringhe.
    """
    processed = set()
    if os.path.exists(processed_ids_path):
        with open(processed_ids_path, 'r', encoding='utf-8') as f:
            for line in f:
                processed.add(line.strip())
    return processed


def append_processed_email(processed_ids_path, email_id):
    """
    Aggiunge l'ID di una email al file dei processati.
    """
    with open(processed_ids_path, 'a', encoding='utf-8') as f:
        f.write(email_id + "\n")


def get_next_number(save_path):
    """
    Scansiona la directory di destinazione e trova il numero massimo
    usato nei nomi (prefisso nel formato '#### - NomeFile') per poi
    restituire il successivo numero (come intero).
    """
    pattern = re.compile(r'^(\d{4}) - ')
    max_num = 0
    for item in os.listdir(save_path):
        match = pattern.match(item)
        if match:
            num = int(match.group(1))
            if num > max_num:
                max_num = num
    return max_num + 1


def find_outlook_folder(root_folder, folder_name):
    """
    Cerca ricorsivamente fra le cartelle di Outlook, partendo da root_folder,
    quella che ha il nome folder_name. Restituisce l'oggetto folder se trovato,
    altrimenti None.
    """
    if root_folder.Name.lower() == folder_name.lower():
        return root_folder
    for subfolder in root_folder.Folders:
        result = find_outlook_folder(subfolder, folder_name)
        if result is not None:
            return result
    return None


def extract_zip(zip_path, extract_to, password):
    """
    Estrae il file ZIP presente in zip_path nella cartella extract_to
    utilizzando la password (se richiesta). La password viene convertita in bytes.
    """
    try:
        with zipfile.ZipFile(zip_path) as zf:
            for member in zf.infolist():
                try:
                    # zipfile richiede la password in formato bytes
                    zf.extract(member, path=extract_to, pwd=str(password).encode())
                except RuntimeError as e:
                    print(f"Impossibile estrarre {member.filename} da {zip_path}: {e}")
    except Exception as e:
        print(f"Errore nell'aprire il file ZIP {zip_path}: {e}")


def process_nested_zips(directory, password):
    """
    Cerca, in maniera ricorsiva, eventuali file ZIP all'interno della cartella 'directory'.
    Se ne trova, li estrae in una sottocartella (con lo stesso nome del file senza .zip)
    e rimuove il file ZIP originale.
    """
    for root, _, files in os.walk(directory):
        for file in files:
            if file.lower().endswith(".zip"):
                zip_path = os.path.join(root, file)
                folder_name = os.path.splitext(file)[0]
                new_folder = os.path.join(root, folder_name)
                os.makedirs(new_folder, exist_ok=True)
                print(f"Estrazione di ZIP annidato: {zip_path} -> {new_folder}")
                extract_zip(zip_path, new_folder, password)
                os.remove(zip_path)
                process_nested_zips(new_folder, password)


def remove_pdf_password(input_path, output_path, password):
    """
    Apre un file PDF (protetto o meno) e lo salva in output_path senza protezione.
    """
    try:
        reader = PdfReader(input_path)
        if reader.is_encrypted:
            reader.decrypt(password)
        writer = PdfWriter()
        for page in reader.pages:
            writer.add_page(page)
        with open(output_path, 'wb') as f_out:
            writer.write(f_out)
        print(f"PDF decriptato: {input_path}")
    except Exception as e:
        print(f"Errore nella rimozione della password per il PDF {input_path}: {e}")


def remove_office_password(input_path, output_path, password):
    """
    Se il modulo msoffcrypto è disponibile, prova a decriptare un file Office
    (doc, xls, ppt, …) e salvarlo in output_path senza protezione.
    """
    if msoffcrypto is None:
        print("Modulo msoffcrypto non installato: non posso decriptare file Office.")
        shutil.copy(input_path, output_path)
        return

    try:
        with open(input_path, "rb") as f_in:
            office_file = msoffcrypto.OfficeFile(f_in)
            if office_file.is_encrypted():
                office_file.load_key(password=password)
                with open(output_path, "wb") as f_out:
                    office_file.decrypt(f_out)
                print(f"File Office decriptato: {input_path}")
            else:
                shutil.copy(input_path, output_path)
    except Exception as e:
        print(f"Errore nella rimozione della password per il file Office {input_path}: {e}")
        shutil.copy(input_path, output_path)


def remove_password_from_file(file_path, password):
    """
    In base all'estensione del file, se possibile, rimuove la password salvando
    una nuova copia non protetta. Se il file non richiede rimozione o se il tipo non
    è gestito, il file viene lasciato invariato.
    """
    ext = os.path.splitext(file_path)[1].lower()
    tmp_path = file_path + ".tmp"
    if ext == ".pdf":
        remove_pdf_password(file_path, tmp_path, password)
        os.replace(tmp_path, file_path)
    elif ext in ['.doc', '.docx', '.xls', '.xlsx', '.ppt', '.pptx']:
        remove_office_password(file_path, tmp_path, password)
        os.replace(tmp_path, file_path)
    else:
        # Tipi non gestiti: non fare nulla
        pass


def process_email(email, config):
    """
    Processa tutti gli allegati di una email:
      - Salva l'allegato aggiungendo un prefisso numerico per evitare duplicati.
      - Se l'allegato è un file ZIP, lo estrae in una cartella (ed eventuali ZIP annidati)
      - Altrimenti, prova a rimuovere la password (ad es. per PDF o file Office).
    """
    save_path = config["SAVE_PATH"]
    password = config["PASSWORD"]

    for i in range(1, email.Attachments.Count + 1):
        attachment = email.Attachments.Item(i)
        next_num = get_next_number(save_path)
        original_name = attachment.FileName
        new_name = f"{next_num:04d} - {original_name}"
        full_path = os.path.join(save_path, new_name)
        try:
            attachment.SaveAsFile(full_path)
            print(f"Allegato salvato: {full_path}")
        except Exception as e:
            print(f"Errore nel salvataggio dell'allegato {original_name}: {e}")
            continue

        ext = os.path.splitext(full_path)[1].lower()
        if ext == ".zip":
            folder_name = os.path.splitext(new_name)[0]
            folder_path = os.path.join(save_path, folder_name)
            os.makedirs(folder_path, exist_ok=True)
            print(f"Estrazione del file ZIP: {full_path} -> {folder_path}")
            extract_zip(full_path, folder_path, password)
            os.remove(full_path)
            process_nested_zips(folder_path, password)
            for root, _, files in os.walk(folder_path):
                for file in files:
                    fpath = os.path.join(root, file)
                    remove_password_from_file(fpath, password)
        else:
            remove_password_from_file(full_path, password)


def main():
    # Percorso del file Excel di configurazione
    CONFIG_FILE = r"C:\Users\Balestra\Desktop\test\dati.xlsx"
    
    # Legge la configurazione
    config = read_config(CONFIG_FILE)
    
    # Crea la cartella di salvataggio se non esiste
    save_path = config["SAVE_PATH"]
    os.makedirs(save_path, exist_ok=True)
    
    # Legge gli ID delle email già processate
    processed_ids_path = config["PROCESSED_EMAILS_PATH"]
    processed_ids = get_processed_email_ids(processed_ids_path)

    # Connessione a Outlook utilizzando DispatchEx
    try:
        outlook_app = win32com.client.DispatchEx("Outlook.Application")
        outlook = outlook_app.GetNamespace("MAPI")
        print("Outlook connesso correttamente.")
    except Exception as e:
        print("Errore nell'apertura di Outlook:", e)
        return

    # Se è specificato EMAIL_ACCOUNT, cerca lo store corrispondente
    if "EMAIL_ACCOUNT" in config and config["EMAIL_ACCOUNT"]:
        account_name = config["EMAIL_ACCOUNT"]
        print("Cercando account:", account_name)
        selected_store = None
        for store in outlook.Folders:
            print("Verifica store:", store.Name)
            if store.Name.lower() == account_name.lower():
                selected_store = store
                break
        if selected_store is None:
            print(f"Account '{account_name}' non trovato. Uso il folder predefinito (Posta in arrivo).")
            root_folder = outlook.GetDefaultFolder(6)
        else:
            print(f"Account trovato: {selected_store.Name}")
            root_folder = selected_store
    else:
        root_folder = outlook.GetDefaultFolder(6)

    # Cerca la cartella target specificata dalla variabile OUTLOOK_SUBFOLDER
    target_folder = find_outlook_folder(root_folder, config["OUTLOOK_SUBFOLDER"])
    if target_folder is None:
        print(f"Cartella Outlook '{config['OUTLOOK_SUBFOLDER']}' non trovata.")
        return
    else:
        print(f"Cartella target trovata: '{target_folder.Name}'")

    print(f"Inizio elaborazione email in '{target_folder.Name}'...")
    for item in target_folder.Items:
        # Verifica che l'oggetto sia una MailItem (classe 43 in Outlook)
        if item.Class != 43:
            continue
        email_id = item.EntryID
        if email_id in processed_ids:
            print("Email già processata, salto...")
            continue
        print(f"Processo email: {item.Subject}")
        process_email(item, config)
        append_processed_email(processed_ids_path, email_id)
    
    print("Elaborazione terminata.")


if __name__ == '__main__':
    main()
```
