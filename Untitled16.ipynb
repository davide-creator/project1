```python
import os
import datetime
import requests  # se devi scaricare il file via HTTP
import openpyxl
from openpyxl.utils import get_column_letter

def scarica_excel_da_sito(url_download, percorso_locale):
    """
    Esempio di funzione che scarica un Excel da un URL e lo salva in locale.
    Se hai già il file in locale, puoi saltare questa parte.
    """
    response = requests.get(url_download)
    with open(percorso_locale, 'wb') as f:
        f.write(response.content)

def elabora_excel(percorso_input, percorso_output, data_oggi=None):
    """
    1) Carica il file Excel scaricato.
    2) Rinomina il workbook come "fund_data_report + data" (se necessario).
    3) Seleziona lo sheet 'Main'.
    4) Aggiunge in coda alcune colonne con formule, come da specifiche.
    5) Salva come 'Pricing Grid' (o nome specificato) in percorso_output.
    """

    # Se data_oggi non è fornita, usiamo la data attuale
    if not data_oggi:
        data_oggi = datetime.date.today().strftime('%Y%m%d')
    # Se vuoi un formato diverso di data, modifica lo strftime.

    # -----------------------------
    # Caricamento workbook
    # -----------------------------
    wb = openpyxl.load_workbook(percorso_input)

    # (Opzionale) se vuoi rinominare il file in memoria, in realtà non serve,
    # basta solo che quando salvi utilizzi un certo nome.
    # Tuttavia, se vuoi proprio salvare un "fund_data_report + Data.xlsx", lo farai dopo.

    # Seleziona lo sheet "Main"
    try:
        sheet = wb["Main"]
    except KeyError:
        raise ValueError("Non esiste un foglio di nome 'Main' nel tuo file Excel.")

    # -----------------------------
    # Eventuale filtro sulle date (colonna BM)
    # ESEMPIO: scartare le righe con data < 2023
    # -----------------------------
    # Supponiamo che la colonna BM sia indice 65 (A=1, ..., BM=65).
    # In openpyxl le colonne si riferiscono in base letterale ("BM") oppure con get_column_letter.
    # Se vuoi controllare la colonna BM per ogni riga:
    #
    #  for row in range(2, sheet.max_row + 1):
    #      cell_data = sheet["BM" + str(row)].value
    #      if isinstance(cell_data, datetime.datetime):
    #          # Se la data è più vecchia di 2023, puoi fare skip, cancellare riga, ecc.
    #          if cell_data.year < 2023:
    #              ...
    #
    # Attenzione che cancellare righe mentre si itera può essere scomodo;
    # meglio collezionare gli indici e poi eliminarli in un secondo momento, se serve.

    # ---------------------------------------------------------------
    # Aggiunta delle nuove colonne in coda con le formule richieste.
    # ---------------------------------------------------------------
    #
    # 1) Trova la prima colonna libera (dopo le esistenti)
    #    Puoi farlo in vari modi. Ad esempio:
    last_col = sheet.max_column  # indice numerico dell'ultima colonna occupata
    start_newcol = last_col + 1  # la prossima colonna libera

    # 2) Creiamo un elenco di "nuove colonne", ciascuna con:
    #    - un nome (header)
    #    - una formula di esempio che si applica dalla riga 2 in giù
    #
    #    NOTA: qui devi adattare la formula effettiva dal testo che hai tu.

    nuove_colonne = [
        {
            "header": "Annual Charge",
            # Esempio: somma di 3 colonne (supponiamo C, D, E) => =C2+D2+E2
            # Metti la formula in inglese (Excel standard).
            "formula_base": "=C{r}+D{r}+E{r}"
        },
        {
            "header": "Check Class Code",
            # Esempio formula (in inglese) che controlla se in colonna A c'è la lettera "C"
            # (equivalente a TROVA/SE.ERRORE in italiano).
            "formula_base": '=IF(ISERROR(FIND("C",A{r})),"No C","Has C")'
        },
        {
            "header": "Derived Accrual Method",
            # Esempio che, se in colonna F c'è "A", scriviamo "Fixed TER", altrimenti "CAP"
            "formula_base": '=IF(F{r}="A","Fixed TER","CAP")'
        },
        {
            "header": "Expected-EOD",
            # Esempio fittizio.
            "formula_base": '=IF(TRIM(A{r})="XYZ","True","C2+S2")'
        },
        {
            "header": "Annual Charge Check",
            # Esempio: differenza fra "Annual Charge" (nuova colonna) e colonna T
            # (per dire se coincide).
            "formula_base": "=({annual_col}{r}) - T{r}"
        },
        {
            "header": "Tax d’Abbo Within Range for Active",
            # Esempio fittizio:
            "formula_base": '=IF(AND(Q{r}>=0.2,Q{r}<=0.8), "OK", TRIM(L{r}))'
        }
    ]

    # 3) Scriviamo l’header di ogni nuova colonna e le formule riga per riga
    for i, col_info in enumerate(nuove_colonne):
        # La colonna di destinazione (indice)
        col_index = start_newcol + i

        # La lettera della colonna (es. se col_index=10 => "J")
        col_letter = get_column_letter(col_index)

        # Scriviamo l'intestazione in riga 1
        sheet[f"{col_letter}1"] = col_info["header"]

        # Ora, per tutte le righe (dalla 2 fino a max_row), inseriamo la formula
        for row in range(2, sheet.max_row + 1):
            # Se per caso vuoi saltare righe vuote, puoi fare un check
            # Esempio: if not sheet[f"A{row}"].value: continue

            # Caso speciale: se la formula fa riferimento a una colonna definita poco fa
            # (tipo "Annual Charge" in colonna?), possiamo sostituire con la lettera appena generata:
            if "{annual_col}" in col_info["formula_base"]:
                # Cerchiamo la colonna dove abbiamo messo "Annual Charge"
                # Supponiamo che "Annual Charge" sia la prima colonna del blocco,
                # quindi start_newcol ha l’indice di questa colonna.
                annual_charge_col_letter = get_column_letter(start_newcol)
                formula = col_info["formula_base"].replace(
                    "{annual_col}", annual_charge_col_letter
                )
            else:
                formula = col_info["formula_base"]

            # Rimpiazziamo {r} con il numero di riga attuale
            formula = formula.replace("{r}", str(row))

            # Inseriamo la formula
            sheet[f"{col_letter}{row}"] = formula

    # -----------------------------
    # Salvataggio finale
    # -----------------------------
    # Esempio: salviamo con nome "Pricing Grid <data_oggi>.xlsx"
    # (oppure semplicemente "Pricing Grid.xlsx")
    nome_file_output = os.path.join(
        percorso_output,
        f"Pricing Grid {data_oggi}.xlsx"
    )
    wb.save(nome_file_output)
    print(f"File salvato correttamente come: {nome_file_output}")


if __name__ == "__main__":

    # ESEMPIO di utilizzo:
    # 1) Scarico il file (se necessario)
    # scarica_excel_da_sito(
    #     url_download="https://esempio.com/fund_data_report.xlsx",
    #     percorso_locale="C:/percorso/locale/fund_data_report.xlsx"
    # )

    # 2) Elabora il file e crea la Pricing Grid
    percorso_input  = r"C:\percorso\locale\fund_data_report.xlsx"
    percorso_output = r"C:\percorso\destinazione"

    # Puoi passare una data specifica, es: "20250615",
    # oppure lasciare vuoto e userà la data di sistema.
    elabora_excel(percorso_input, percorso_output, data_oggi="20250615")
```
